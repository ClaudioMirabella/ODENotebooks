```python
FIGURE_SIZE = (7.0, 3.0)
N_POINTS = 200
RATE_RANGE = (-1.0, 1.0, 0.1)               # minimum, maximum, step
DEFAULT_RATE = 0.5


def plot_exponential(k=DEFAULT_RATE):
    # the function exp(k x) for 0 <= x <= 3
    points = np.linspace(0.0, 3.0, N_POINTS)
    fig, ax = plt.subplots(figsize=FIGURE_SIZE)
    ax.plot(points, np.exp(k * points), "-", linewidth=2)
    ax.set_ylim(0.0, np.exp(RATE_RANGE[1] * 3.0))
    ax.set_xlabel("x")
    ax.set_title(f"exp(k x) with k = {k:g}")
    plt.show()


interact(plot_exponential,
         k=FloatSlider(value=DEFAULT_RATE, min=RATE_RANGE[0], max=RATE_RANGE[1], step=RATE_RANGE[2], description="k",
                       continuous_update=False));
```

## 3. How to type mathematics

| On paper | In Python | The usual mistake |
|---|---|---|
| 3.5 | `3.5` | 3,5 (a comma: Python reads the two numbers 3 and 5) |
| 2 to the power 10 | `2 ** 10` | 2^10 (no red message with whole numbers: Python computes something else, here 8) |
| 3x,  2(x + 1) | `3 * x,  2 * (x + 1)` | 3x,  2(x + 1) (the product is always written) |
| 3/4 | `3 / 4` |  |
| 1 over 2x,  x to the power 1/2 | `1 / (2 * x),  x ** (1 / 2)` | 1 / 2 * x,  x ** 1 / 2 (brackets around a denominator and an exponent) |
| one third, the square root of 2 | `1 / 3,  np.sqrt(2)` | 0.33,  1.41 (type the exact expression, not a rounded number) |
| square root of 2, in a formula | `sp.sqrt(2)` | sqrt(2) |
| e to the power -2x | `np.exp(-2 * x)` or, in a formula, `sp.exp(-2 * x)` | e**(-2x),  exp(-2 * x) |
| ln x | `np.log(x)` or, in a formula, `sp.log(x)` | ln(x) |
| sin x, cos x | `np.sin(x), np.cos(x)` or `sp.sin(x), sp.cos(x)` | sin x,  sin(x),  sen(x) |
| pi | `np.pi` or, in a formula, `sp.pi` | pi |
| a list of numbers | `[0, 2, 1]` | 0 2 1 |
| yes, no | `True, False` | true, yes, "True" (no quotes) |
| infinitely many | `np.inf` | infinity |

np. is for numbers, sp. is for formulas that contain a letter such as x or t: an exercise that asks for a formula says so, and names the letter. Inside a function (a line that starts with return) and in plots use np.: there the letter stands for numbers.

## 4. When you see red

A red box is an error message, not a disaster. Read its **last line**, and the line that starts with *Hint (MATH 331)* just above or below the box, when there is one.

| The last line of the red message | What it means | What to do |
|---|---|---|
| `NameError: name 'np' is not defined` (or 'ol', 'sp', 'plt', 'x', ...) | The cell was run before the cells above it. | Run the whole notebook: the >> button, or Run > Run All Cells. |
| `NameError: name 'exp' is not defined` (or 'sqrt', 'sin', 'pi', 'e', ...) | A function or a constant was typed without np. or sp. in front. | Write np.exp(2) with numbers, sp.exp(x) in a formula. See 'How to type mathematics'. |
| `NameError: name 'my_x' is not defined` | The line my_x = ... was deleted, or was typed with two = signs. | Type it again above the line # ----, with one = sign: my_x = None. |
| `NameError: name '...' is not defined` (any other name) | A name is used that Python does not know: often a typing mistake. | Check the spelling, capital letters included. |
| `SyntaxError: invalid decimal literal` | A product without its sign (3x, 2x**2), or a unit after a number (3.5m). | Write 3 * x; type the number without its unit. |
| `SyntaxError: invalid syntax` or '(' was never closed | A bracket is missing, nothing is typed after =, or something that is not Python was typed (a word, %, :). | Look at the place marked with ^ under the line, and at the line before it. |
| `SyntaxError: invalid character ...` | A sign was pasted from a document: a long minus, a multiplication cross, a curly quote. | Type the signs - * / and " on the keyboard. |
| `IndentationError: unexpected indent` | A line starts with a space. | Lines start at the left edge; only the lines inside def or for are moved right, by four spaces. |
| `TypeError: 'int' object is not callable` (or 'float', 'Symbol', ...) | A product without its sign before a bracket: 2(3 + 1). (Ignore the words about a comma that Python adds.) | Write 2 * (3 + 1). |
| `TypeError: unsupported operand type(s) for ^` | A power was typed with ^. (With whole numbers there is no message, only a wrong value.) | Write two stars for a power: x ** 3. |
| `TypeError: loop of ufunc does not support argument ...` | A np. function was used in a formula with a letter. | In a formula use sp.exp, sp.sin, sp.sqrt, ... |
| `AssertionError` (nothing after it) | You changed a number in an example. The cell compares its result with a value of the book, and they no longer agree. Nothing is broken. | Put the old number back, or delete the line that starts with assert and go on. |
| `AssertionError: some words` | In a 'Fill in' exercise: the function you completed does not give the right values yet; the words say what failed. | Correct the line return ... in the cell above, run that cell, then this one again. |
| `... : not yet (...)` | This is not an error: the cell ran, and your answer is not the expected one. | Read the words in brackets and try again. |
| `A cell shows [*] and never finishes` | Python is busy, or stuck. (The first time you open the site, a minute or two of [*] is normal: wait.) | Menu Kernel > Restart Kernel, then run everything again. The square button may not stop it in the browser. |

Still stuck? Copy the last line of the message and bring it to class or to office hours.

## 9. For later: conventions of the notebooks

* **Letters.** The unknown is $y(x)$ when an equation is studied for itself, as in the course text;
  it is a function of the time $t$ in the models ($x(t)$, $v(t)$, $T(t)$, $h(t)$).
* **Integrating factor.** $r(x)$ for a linear equation (Lebl, Section 1.4) and $\mu$ for an exact
  equation: the first is a special case of the second (notebook 12).
* **Logistic equation.** $x' = kx(M - x)$, as in Lebl, Section 1.6.
* **Gravity.** $g = 9.81$ m/s$^2$ everywhere (`ol.GRAVITY`).
* **Exercise numbers** of the course text are those of the lists at the end of each chapter of the
  LibreTexts edition, unless "in the text" is said.
* **Checks.** `ol.check` compares a number, `ol.check_formula` a formula typed with SymPy,
  `ol.check_solution` substitutes a formula in an equation; all answer *correct* or *not yet* and
  never show the answer.
* **Self-checks.** Worked examples are compared with the values printed in the books by lines that
  start with `assert`: the notebook stops if a number does not match.
